```python
import pandas as pd
from pathlib import Path
import re

# =========================================================
# PATHS
# =========================================================

LST_DIR = Path("../data/state")
OUTPUT_DIR = Path("../data/variables")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_CSV = OUTPUT_DIR / "lst_raster.csv"

# =========================================================
# FIND ALL LST FILES
# =========================================================

pattern = re.compile(r"LST_ASSAM_\d{4}_\d{2}\.tif")

lst_files = [
    f.name for f in LST_DIR.iterdir()
    if f.is_file() and pattern.match(f.name)
]

lst_files = sorted(lst_files)

# =========================================================
# CREATE DATAFRAME
# =========================================================

df = pd.DataFrame({
    "land-surface-temperature-raster": lst_files
})

# =========================================================
# SAVE CSV
# =========================================================

df.to_csv(OUTPUT_CSV, index=False)

print("Saved:", OUTPUT_CSV)
print("Total rasters found:", len(df))
print(df.head())
```
